```python
import asyncio
from websockets.asyncio.server import serve
```

```python
# Then run it from anywhere. One caveat: the watch app has 10.5.13.122 hardcoded as the server IP (WebSocketClient.kt:31), so you'll either need to:
# - Set your PC's IP to 10.5.13.122, or
# - Change that IP in the Kotlin source and re-deploy the watch app

WATCH_CONNECTED = None

async def handler(websocket):
    global WATCH_CONNECTED
    WATCH_CONNECTED = websocket
    print("[green]Watch connected![/green]")
    async for message in websocket:
        print(f"Received: {message}")

async def main():
    async with serve(handler, "0.0.0.0", 8765) as server:
        print("Server running on port 8765. Vibrating watch every 10 seconds.")
        while True:
            await asyncio.sleep(10)
            if WATCH_CONNECTED:
                await WATCH_CONNECTED.send("EXECUTE_VIBRATION")
                print("Vibration sent!")
                #TODO: need to add an osc message here with an id so that we can log the vibration and response on both ends
            else:
                print("No watch connected yet.")

asyncio.run(main())
```
